```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train-data.csv')
df.head()
```

```python
df.isna().sum()
```

```python
df.isna().mean()*100
```

```python
df.drop(columns=['New_Price','Unnamed: 0'],inplace=True)
```

```python
df.ffill(inplace=True)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
data = ohe.fit_transform(df[['Location']])
df1 = pd.DataFrame(data,columns=ohe.get_feature_names_out())
```

```python
ohe1=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
data1=ohe1.fit_transform(df[['Fuel_Type']])
df2=pd.DataFrame(data1,columns=ohe1.get_feature_names_out())
```

```python
ohe2=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
data2=ohe2.fit_transform(df[['Transmission']])
df3=pd.DataFrame(data2,columns=ohe2.get_feature_names_out())
```

```python
df=pd.concat([df,df1,df2,df3],axis=1)
```

```python
df.drop(['Fuel_Type','Transmission','Location'],axis=1,inplace=True)
```

```python
df.info()
```

```python
df['Owner_Type'].unique()
```

```python
df['Owner_Type']=df['Owner_Type'].replace({'First':0, 'Second':1, 'Fourth & Above':3, 'Third':2})
df['Owner_Type']=df['Owner_Type'].astype(int)
```

```python
df['Engine'] = df['Engine'].replace({'CC' :''},regex=True)
df['Engine'] = df['Engine'].astype(int)
```

```python
for i in df['Power'].unique():
    if not i.endswith('bhp'):
      print(i)
```

```python
df['Power'] = df['Power'].replace({'bhp' :''},regex=True)
```

```python
df['Power']=df['Power'].str.strip()
```

```python
import re
for i in df['Power'].unique():
    if not re.match(r'^[0-9,]+$',i):
        print(i)
```

```python
df['Power'] =  df['Power'].replace({'null':0})
df['Power'] = df['Power'].astype(float)
```

```python
for i in df['Mileage'].unique():
    if not i.endswith('km/kg') and not i.endswith('kmpl'):
      print(i)
```

```python
def conv_Mileage(row):
    row = row.lower().strip()
    if row.endswith('km/kg'):
        num = row.replace('km/kg','')
        return float(num)*1.4
    elif row.endswith('kmpl'):
         num = row.replace('kmpl','')
         return float(num)
        
df['Mileage'] = df['Mileage'].apply(conv_Mileage)
```

```python
df.info()
```

```python
df['Power']=df['Power'].replace({0:np.nan})
```

```python
df.isna().sum()
```

```python
df.ffill(inplace=True)
```

```python
df.isna().sum()
```

```python
x=df.drop(['Price'],axis=1)
y=df['Price']
```
